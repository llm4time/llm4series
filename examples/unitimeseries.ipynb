# LLM4Series — Visualizing a Univariate Time Series

This notebook demonstrates the built-in plotting capabilities of a `UniTimeSeries` object.
All plots are produced through the same `.plot()` method by changing the `kind` parameter.

## Step 1 — Import the library and load the data

`OT.csv` has a single variable, so `read_file` returns a `UniTimeSeries`.

```python
import llm4series as ls

ts = ls.read_file("data/OT.csv", index_col="date")
ts
```

## Step 2 — Interactive chart plot

`kind='chart'` produces an interactive time series overview with zoom and pan controls.

```python
fig = ts.plot(kind="chart", title="OT", xlabel="Date", ylabel="Value")
fig.show(renderer="svg")
```

## Step 3 — Line plot on a slice

Slice the series with `[:100]` to inspect only the first 100 observations.
`kind='line'` draws a simple line chart, useful for spotting local patterns.

```python
fig = ts[:100].plot(kind="line", title="OT", xlabel="Date", ylabel="Value")
fig.show(renderer="svg")
```

## Step 4 — Bar / statistics plot

`kind='bar'` displays descriptive statistics (mean, std, min, max, etc.)
as a bar chart, giving a quick distributional summary of the entire series.

```python
fig = ts.plot(kind="bar", title="OT", xlabel="Statistics", ylabel="Value")
fig.show(renderer="svg")
```

## Step 5 — STL decomposition plot

`kind='stl'` decomposes the series into trend, seasonal, and residual components
using Seasonal-Trend decomposition with LOESS. Useful for understanding seasonality.

```python
fig = ts.plot(kind="stl", title="OT", xlabel="Date", ylabel="Value")
fig.show(renderer="svg")
```

## Step 6 — Autocorrelation plot

`kind='autocorr'` plots the autocorrelation function (ACF) across lags.
It reveals whether past values are correlated with future values,
which is important for choosing forecasting models.

```python
fig = ts.plot(kind="autocorr", title="OT", xlabel="Lag", ylabel="Value")
fig.show(renderer="svg")
```
