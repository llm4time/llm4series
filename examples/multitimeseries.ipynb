# LLM4Series — Visualizing a Multivariate Time Series

This notebook demonstrates the plotting capabilities of a `MultiTimeSeries` object.
The same `kind` options used for univariate series apply here, but each plot
now covers all variables simultaneously.

## Step 1 — Import the library and load the data

`ETTh2.csv` contains seven variables (HUFL, HULL, MUFL, MULL, LUFL, LULL, OT),
so `read_file` automatically returns a `MultiTimeSeries`.

```python
import llm4series as ls

ts = ls.read_file("data/ETTh2.csv", index_col="date")
ts
```

## Step 2 — Interactive chart plot

`kind='chart'` renders an interactive multi-line overview of all seven variables.
Each variable gets its own color and can be toggled in the legend.

```python
fig = ts.plot(kind='chart', title="ETTh2 Dataset", xlabel="Date", ylabel="Value")
fig.show(renderer='png')
```

## Step 3 — Line plot on a slice

Slicing with `[:100]` limits the view to the first 100 timestamps.
`kind='line'` draws all variables as overlapping lines, which helps
compare short-term behavior across variables.

```python
fig = ts[:100].plot(kind='line', title="ETTh2 Dataset", xlabel="Date", ylabel="Value")
fig.show(renderer='png')
```

## Step 4 — Bar / statistics plot

`kind='bar'` shows descriptive statistics for each variable side by side.
This makes it easy to compare the scale and spread of different variables
before feeding them into a model.

```python
fig = ts.plot(kind='bar', title="ETTh2 Dataset", xlabel="Statistics", ylabel="Value")
fig.show(renderer='png')
```

## Step 5 — STL decomposition plot

`kind='stl'` runs STL decomposition on every variable and displays
the trend, seasonal, and residual components in a grid layout,
one row per variable.

```python
fig = ts.plot(kind='stl', title="ETTh2 Dataset", xlabel="Date", ylabel="Value")
fig.show(renderer='png')
```
